```python
import torch
```

```python
class BitTensor(torch.Tensor):

    def __new__(cls, data, *, requires_grad=False):
        data = torch.as_tensor(data, dtype=torch.bool)
        return torch.Tensor._make_subclass(cls, data, requires_grad)

    def __repr__(self):
        return f"BitTensor({super(BitTensor, self).__repr__()})"
    
    @classmethod
    def from_tensor(cls, x):
        return cls(x)

    def to_tensor(self):
        return self.as_subclass(torch.Tensor)
    
    def int(self):
        return self.to_tensor().to(torch.int)
    

    def __and__(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() & other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for &: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def __or__(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() | other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for |: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def __xor__(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() ^ other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for ^: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def __invert__(self):
        return BitTensor(~self.to_tensor())
    
    def logical_and(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() & other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for logical_and: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def logical_or(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() | other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for logical_or: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def logical_xor(self, other):
        if isinstance(other, BitTensor):
            return BitTensor(self.to_tensor() ^ other.to_tensor())
        return NotImplementedError("Unsupported operand type(s) for logical_xor: 'BitTensor' and '{}'".format(type(other).__name__))
    
    def logical_not(self):
        return BitTensor(~self.to_tensor())
```
